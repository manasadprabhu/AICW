```python
from datetime import date
```

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year - birthyear)

    def show(self):
        print(self.name + "'s age is "+str(self.age))
```

```python
age = student('Manasa',20)
age.show()
```

```python
by=student.calculate_age('Manasa',2006)
by.show()
```

```python
class School:
    #class variable
    name="BGS"
    @classmethod                #classmethod allows a method to access and modify class variables.
    def school_name(cls):
        print('School name is ', cls.name)
    #create class method
    School.school_name = classmethod(School.school_name)

    #call class method
School.school_name()
```

```python
#MINI PROJECT
class BankAccount:
    def __init__(self,name,balance=0):
        #instance attributes
        self.name=name
        self.balance=balance
    #deposit money
    def deposit(self,amount):        #instance method
        if amount>0:
            self.balance += amount
            print("Deposit Successfully!")
            print("Deposited:",amount)

        else:
            print("Invalid deposit amount")

    #withdraw money
    def withdraw(self,amount):
        if amount <=0:
            print('Invalid withdrawl amount!')
        elif amount > self.balance:
            print('Insufficient balance!')
        else:
            self.balance -= amount
            print("Withdrawl successfully!")
            print("Withdrawn:",amount)
    

        #check balance
    def check_balance(self):
        print("Current balance:",self.balance)
        #display account details
    def display_details(self):
        print("Account Holder:",self.name)
        print("Balance:",self.balance)
    
```

```python
acc1=BankAccount('Manasa',2000)
acc1.display_details()
```

```python
acc1.deposit(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(2000)
```

```python
acc1.check_balance()
```

# Inheritance,polymorphism and encapsulation

```python
#parent class
class Person:
    def person_info(self,name,age):
        print('Inside Person class')
        print('Name:',name,'Age:',age)
#parent class 2
class Company:
    def company_info(self,company_name,location):
        print('Inside the Company class')
        print('Name:',company_name,'Location:',location)
```

```python
#child class
class Employee(Person,Company):
    def Employee_info(self,salary,skill):
        print('Inside employee class-child class')
        print('Salary',salary, 'Skills:',skill)
```

```python
#create an object of Employee
emp2 = Employee()
```

```python
#access the data
emp2.person_info('Manasa',20)
emp2.company_info('Edunet Foundation','Bengaluru')
emp2.Employee_info(50000,'AIML Engineer')
```

```python
#super - function
class Company:
    def company_name(self):
        return 'Edunet Foundation'
class Employee(Company):
    def info(self):
        #call the super func
        c_name = super().company_name()
        print('Manasa works at', c_name)
```

```python
#METHOD OVERRIDING
class Vehicle:
    def max_speed(self):
        print("max speed is 100 kmph")
class Car(Vehicle):
    #overriding the implentation of vehicle class
    def max_speed(self):
        print('max speed is 200 kmph')
```

```python
#public member
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    #public instance method
    def show(self):
        #access public data members
        print('Name',self.name,'Salary',self.salary)
```

```python
emp = Employee('Manasa',50000)
#access the public data members
print('Name',emp.name,'Salary',emp.salary)
emp.show()
```

```python
#private member
class Employee:
    def __init__(self,name,salary):
        self.name=name #public member
        self.__salary=salary  #private member
```

```python
emp=Employee('Manasa',50000)
#access the private data member
print('Salary',emp.__salary)
```

```python
# protected member
class Company:
    def __init__(self):
        #protected member
        self._project="NLP"       #protected = _     private = __
#child class
class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)
    def show(self):
        print("Employee Name:",self.name)
        #access the protected member
        print('Working on project:',self._project)
```

```python
c = Employee("Manasa")
c.show()
```

```python
#direct access
print('Project:',c._project)
```

```python
# polymorphism with inheritance
class Bird:
    def intro(self):
        print('There are many types of birds!')
    def flight(self):
        print('Most of the birds can fly but some cannot')

class sparrow(Bird):
    def flight(self):
        print("Sparrows can fly!")

class ostrich(Bird):
    def flight(self):
        print("Ostrich cannot fly")
```

```python
ob_bird=Bird()
ob_bird.intro()
ob_bird.flight()
```

```python
ob_sparrow=sparrow()
ob_sparrow.flight()
```

```python
ob_ostrich=ostrich()
ob_ostrich.flight()
```

```python
#overloading
class complex:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def __add__(self,other):
        return self.a+other.a, self.b+other.b
ob1=complex(1,2)
ob2=complex(2,3)
ob3=ob1+ob2
print(ob3)
```
